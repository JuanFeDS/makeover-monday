```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go
```

## **Datos**

```python
df = pd.read_excel("./data/MM 2026wk15.xlsx")
df
```

```python
df.columns = [col.replace(" ", "_").lower() for col in df.columns]
```

```python
df.rename({
    'layoffs_(past_2yr)_-_people': 'layoffs_people',
    'layoffs_(past_2yr)_-_events': 'layoffs_events'
    }, axis=1, inplace=True)
df
```

```python
df['%_open_jobs'] = df['open_jobs'].apply(lambda x: round(x / df['open_jobs'].sum()*100, 2))
df['%_employees'] = df['employees'].apply(lambda x: round(x / df['employees'].sum()*100, 2))
df['%_layoffs_people'] = df['layoffs_people'].apply(lambda x: round(x / df['layoffs_people'].sum()*100, 2))
```

```python
df['%_open_jobs_company'] = round(df['open_jobs'] / df['employees'] * 100, 2)
df['%_layoffs_people_company'] = round(df['layoffs_people'] / df['employees'] * 100, 2)
df.sort_values('employees', ascending=False)
```

```python
df['diff'] = df['%_layoffs_people_company'] - df['%_open_jobs_company']
```

```python
df.to_csv('./data/MM_2026wk15.csv', index=False)
```
