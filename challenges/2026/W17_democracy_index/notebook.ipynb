## **Datos**

```python
df = pd.read_excel('./data/2024 Democracy Index.xlsx')
```

```python
df.columns = [col.replace(' ', '_').lower() for col in df.columns]
```

```python
df.sample(5)
```

```python
df.groupby('year').count()[['entity']]
```

```python
df.query('year==2024').groupby('world_region_according_to_owid').agg({
    'entity': ['count'],
    'democracy_index': ['mean', 'median', 'min', 'max']
})
```

```python
df.query('year==2006').groupby('world_region_according_to_owid').agg({
    'entity': ['count'],
    'democracy_index': ['mean', 'median', 'min', 'max']
})
```

```python
index_by_region_and_year = df.groupby(['world_region_according_to_owid', 'year']).agg({
    'democracy_index': ['mean', 'median']
}).reset_index()
```

```python
index_by_region_and_year.to_csv('data/index_by_region_and_year.csv', index=False)
```
