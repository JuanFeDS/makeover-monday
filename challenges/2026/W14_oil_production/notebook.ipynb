## **Librerías**

```python
import pandas as pd
import numpy as np
```

```python
import plotly.express as px
import plotly.graph_objects as ga
```

## **Datos**

```python
df = pd.read_csv('data/oil-production-by-country.csv')
```

```python
df.sample(5)
```

```python
df.groupby(['Entity'])["Year"].count()
```

```python
df_resume = df[df['Year'].isin([1960, 1970, 1980, 1990, 2000, 2010, 2020, 2026])].copy()
```

```python
df_resume.query('Entity == "Norway"')
```

```python
df_last_year = df.query('Year == 2020')
df_last_year
```

```python
df_last_year['%_Oil'] = df_last_year['Oil'].apply(lambda x: round(x/df_last_year['Oil'].sum() * 100, 2))
df_last_year.sort_values('%_Oil', ascending=False)
```

```python
df_norway = df_resume[df_resume["Entity"] == "Norway"]

fig = px.line(
    df_resume,
    x="Year",
    y="Oil",
    markers=True,  # 👈 esto agrega los puntos
    title="Oil Production Over Time - Norway"
)

fig.show()
```

```python
df_norway = df[df["Entity"] == "Norway"]

fig = px.line(
    df_norway,
    x="Year",
    y="Oil",
    markers=True,  # 👈 esto agrega los puntos
    title="Oil Production Over Time - Norway"
)

fig.show()
```
